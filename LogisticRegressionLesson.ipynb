```python
import matplotlib.pyplot as plt
# `load_digits` will give me a dummy digits data
from sklearn.datasets import load_digits
from sklearn.linear_model import LogisticRegression
from sklearn.model_selection import train_test_split
```

```python
X = load_digits().data
y = load_digits().target
```

```python
x_train, x_test, y_train, y_test = train_test_split(X, y, test_size=0.5)
```

```python
# `imshow` taskes an argument in two dimension array So, here we reshaped the `X`
plt.imshow(X[12].reshape(8, 8))
```

```python
# `fit_intercept` means that my equation contains bias, `solver` argument means that the algorithm which you want to use, `liblinear` is the defualt you can use `newton-cg` type
# For more info visit: https://stackoverflow.com/q/46779605/11019205
model = LogisticRegression(n_jobs=1, fit_intercept=True, solver='liblinear')
model.fit(x_train, y_train)
```

```python
model.score(x_test, y_test)
```

```python
from sklearn.metrics import accuracy_score
```

```python
y_pred = model.predict(x_test)
```

```python
accuracy_score(y_pred=y_pred, y_true=y_test)
```
