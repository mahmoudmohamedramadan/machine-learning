```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import load_iris
```

```python
X = load_iris().data
y = load_iris().target
```

```python
# each circle color points to a different cluster
plt.scatter(X[:, 1], X[:, 3], c=y)
```

```python
X1 = np.linspace(0, 4, 100)
y1 = np.linspace(0, 2.25, 100)
```

```python
'''
Here we draw a line to separate the two clusters BUT we will Note that there are some noises (Circles stays on the line)
'''
plt.scatter(X[:, 1], X[:, 3], c=y)
plt.plot(X1, y1)
```

```python
X2 = np.linspace(1, 4, 100)
y2 = np.linspace(0, 1, 100)
```

```python
plt.scatter(X[:, 1], X[:, 3], c=y)
plt.plot(X1, y1)
plt.plot(X2, y2)
```

```python
# Read about `GridSearchCV` from sklearn
from sklearn.svm import LinearSVC
```

```python
model = LinearSVC(max_iter=4000)
model.fit(X, y)
model.score(X, y)
```
