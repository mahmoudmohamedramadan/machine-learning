```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv('jordan_career.csv')
```

```python
df.head()
```

```python
X = df.iloc[:, [16, 17, 18, 19, 20, 21, 22]]
y = df['pts']
```

```python
from sklearn.preprocessing import MinMaxScaler
```

```python
# Read more about `MinMaxScaler` here: https://vitalflux.com/minmaxscaler-standardscaler-python-examples/
normalization = MinMaxScaler()
X = pd.DataFrame(normalization.fit_transform(X), columns=['orb', 'drb', 'trb', 'ast', 'stl', 'blk', 'tov'])
```

```python
'''
Principal Component Analysis (PCA) is an unsupervised, non-parametric statistical technique primarily used for dimensionality reduction in machine learning
'''
from sklearn.decomposition import PCA
```

```python
model = PCA(n_components=1)
X = model.fit_transform(X)
```

```python
plt.scatter(X, y)
```
