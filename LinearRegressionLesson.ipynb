```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
```

```python
# linspace(start, end, number)
x = np.linspace(0, 5, 101)
b = 5
w = 3

# random(number) and Here we will give an array values of errors
error = np.random.random(101)
y = w * x + b + error
```

```python
# The below line to maximize the scatter size
# plt.figure(figsize=(10, 7))
# `c` refer to the color and `alpha` refer to the color concentration
plt.scatter(x, y, c='red', alpha=0.8, linewidths=2)
```

```python
model = LinearRegression()
```

```python
'''
There are fixed steps in all models are, `fit` to learn the model then `predict` and here the model give me the result depending on what it learns
'''
# If you run the below line will give an error because always the first param(feature) must be in two dimension and the second param(label) must be in one dimension
# model.fit(x, y)
model.fit(x.reshape(-1, 1), y)
```

```python
y_pred = model.predict(x.reshape(-1, 1))
```

```python
plt.scatter(x, y, c='red', alpha=0.8, linewidths=2)
plt.plot(x, y_pred, c='black')
```

```python
from sklearn.metrics import mean_squared_error
```

```python
error = np.sqrt(mean_squared_error(y_pred=y_pred, y_true=y))
error
```
