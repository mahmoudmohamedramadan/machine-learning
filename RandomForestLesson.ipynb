```python
import pandas as pd
```

```python
# If you have a nan values problem you can use `imputer` from sklearn
df = pd.read_csv('winequality-red.csv')
df.head()
```

```python
'''
Seaborn is an open-source Python library built on top of matplotlib. It is used for data visualization and exploratory data analysis
'''
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
# First number is a width and the second is the height
plt.figure(figsize=(12, 5))
# `corr` method Compute pairwise correlation of columns, excluding NA/null values
# `annot` parameter give me the accuracy
sns.heatmap(df.corr(), annot=True)
```

```python
'''
The difference between `loc` and `iloc` loc is label-based, which means that you have to specify rows and columns based on  their row and column labels. iloc is integer position-based, so you have to specify rows and columns by their integer position values
'''
X = df.iloc[:, 0:10]
y = df['quality']
```

```python
from sklearn.model_selection import train_test_split
```

```python
x_train, x_test, y_train, y_test = train_test_split(X, y, test_size=0.2)
```

```python
from sklearn.ensemble import RandomForestClassifier
```

```python
model = RandomForestClassifier(n_jobs=1, n_estimators=100)
```

```python
model.fit(x_train, y_train)
```

```python
model.score(x_test, y_test)
```
